```python
from sqlalchemy import create_engine
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


```

```python
engine = create_engine("mysql+pymysql://root:password@localhost/student_snowflake_dw")

query = """
SELECT dsub.subject_name, AVG(f.g3) AS avg_grade
FROM fact_student_performance f
JOIN dim_subject dsub ON f.subject_id = dsub.subject_id
GROUP BY dsub.subject_name;
"""
```

```python
df = pd.read_sql(query, engine)

sns.barplot(data=df, x="subject_name", y="avg_grade")
plt.title("Average Grade by Subject")
plt.show()
```
